```python
import pandas as pd
```

```python
from pathlib import Path


def load_benchmark_txt(path: Path) -> pd.DataFrame:
    header = [h.strip() for h in path.read_text().splitlines()[0].split(",")]
    df = pd.read_csv(path, skiprows=1, header=None)

    # Some files have more numeric columns than header labels.
    if len(header) < df.shape[1]:
        extra = [f"extra_{i}" for i in range(1, df.shape[1] - len(header) + 1)]
        header = header + extra
    elif len(header) > df.shape[1]:
        header = header[: df.shape[1]]

    df.columns = header
    return df


base = Path("..")


df_1gpu_tile8 = load_benchmark_txt(base / "benchmark_results_1gpu_tile8.txt")
df_1gpu_tile16 = load_benchmark_txt(base / "benchmark_results_1gpu_tile16.txt")
df_1gpu_tile32 = load_benchmark_txt(base / "benchmark_results_1gpu_tile32.txt")
df_2gpu = load_benchmark_txt(base / "benchmark_results_v6_2gpu.txt")

# Quick check
display(df_1gpu_tile8.head())
display(df_1gpu_tile16.head())
display(df_1gpu_tile32.head())
display(df_2gpu.head())
```

```python
df_1gpu_tile16['T_v6(2gpu)'] = df_2gpu['T_v6(2gpu)']
df_1gpu_tile32['T_v6(2gpu)'] = df_2gpu['T_v6(2gpu)']
df_1gpu_tile8['T_v6(2gpu)'] = df_2gpu['T_v6(2gpu)']

col_names = ['N', 'T_seq_v2', 'T_naive', 'T_v1', 'T_v2', 'T_v3(f)', 'T_v5(f)',
       'T_v6(2gpu)', 'Speedup(naive_vs_seq_v2)', 'Speedup(v2_vs_seq_v2)',
       'Speedup(v3_vs_seq_v2)', 'Speedup(v5_vs_seq_v2)',
       'Speedup(v6_vs_seq_v2)']

for df in [df_1gpu_tile8, df_1gpu_tile16, df_1gpu_tile32]:
    if df.shape[1] != len(col_names):
        raise ValueError(f"Column count mismatch: df has {df.shape[1]} columns, expected {len(col_names)}")
    df.columns = col_names
    df['Speedup(v6_vs_seq_v2)'] = df['T_seq_v2'] / df['T_v6(2gpu)']

display(df_1gpu_tile8.head())
display(df_1gpu_tile16.head())
display(df_1gpu_tile32.head())
```

```python
df_1gpu_tile8.to_csv(base / "benchmark_results_1gpu_tile8.csv", index=False)
df_1gpu_tile16.to_csv(base / "benchmark_results_1gpu_tile16.csv", index=False)
df_1gpu_tile32.to_csv(base / "benchmark_results_1gpu_tile32.csv", index=False)
```
